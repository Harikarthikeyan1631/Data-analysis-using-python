```python
# 1. LIST CREATION

#1.a create age list
age_list = [24, 25, 26, 27, 28]
print(age_list)

#1.b create name list
name_list = ["Arun", "Bala", "Charan", "Divya", "Esha"]
print(name_list)

```

```python
#2. LIST OPERATIONS/MODIFICATIONS

# 2.a append "YAZHINI"
name_list.append("Yazhini")
print(name_list)
```

```python
# 2.b INSERT 30 AT INDEX 2
age_list.insert(2, 30)
print(age_list)
```

```python
#2.c Remove the string "Yazhini" from name_list
name_list.remove("Yazhini")
print(name_list)
```

```python
# 2.d Pop the last element from age_list
age_list.pop()
print(age_list)
```

```python
# 2.e Extend the age_list with additional ages [29, 30, 26].
age_list.extend([29, 30, 26])
print(age_list)
```

```python
# 2.f Sort age_list in descending order.
age_list.sort(reverse=True)
print(age_list)
```

```python
# 2.g Find Max age, Min age and sum of all ages from age_list.
max_age = max(age_list)
min_age = min(age_list)
total_age = sum(age_list)
print("Maximum age:", max_age)
print("Minimum age:", min_age)
print("Sum of all ages:", total_age)
```

```python
# 3. Accessing List Elements
# 3.a Print the first element of name_list.
print(name_list[0])
```

```python
# 3.b Print the last element of name_list.
print(name_list[-1])
```

```python
# 3.c Print the elements from index 2 to index 4 in name_list.
print(name_list[2:5])
```

```python
# 3.d Print the elements of name_list in reverse order.
print(name_list[::-1])
```

```python
# 5. SET OPERATIONS
# 5.a Create my_set
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}
print(my_set)
```

```python
#5.b Attempt to Change a Set Element
my_set[4] = 's'
```

```python
# 5.c Create two sets:
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
print("Set 1:", set1)
print("Set 2:", set2)
```

```python
# 5.d Union
union_set = set1.union(set2)
print("Union:", union_set)
```

```python
# 5.e Intersection
intersection_set = set1.intersection(set2)
print("Intersection:", intersection_set)
```
